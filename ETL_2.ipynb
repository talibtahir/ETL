# Data Tranformation

```python
df.isna().sum()
```

```python
df["ROW_ID"] = df["ROW_ID"].astype(int) 
```

```python
df.ORDER_ID
```

```python
df.ORDER_DATE
```

```python
df["ORDER_DATE"] = pd.to_datetime(df["ORDER_DATE"])
```

```python
df["ORDER_DATE"].isna().sum()
```

```python
df["SHIP_DATE"] = pd.to_datetime(df["SHIP_DATE"])
```

```python
df.SHIP_DATE.isna().sum()
```

```python
df["SHIP_MODE"] = df["SHIP_MODE"].astype(str)
```

```python
df.columns 
```

```python
df.POSTAL_CODE = df["POSTAL_CODE"].fillna(0)
```

```python
df["SALES"] = (df["SALES"]
    .str.replace("$", '')
    .str.replace(",", '')
    .astype(float)
)
```

```python
df.QUANTITY = df.QUANTITY.astype(int)
```

```python
df.DISCOUNT = df.DISCOUNT.astype(float)
```

```python

df["PROFIT"] = (df["PROFIT"]
    .str.replace("$", '')
    .str.replace(",", '')
    .astype(float)
    )
```

```python
df['SHIPPING_COST'] = df['SHIPPING_COST'].astype(float)
```

```python
df.head(5)
```

# Data Preparation

```python
df["TOTAL_PRICE"] = (df['SALES'] * df['QUANTITY'])
```

```python
df[['TOTAL_PRICE']] 
```

```python
df['BEST_SELLING'] = (
    df.groupby('PRODUCT_NAME')['QUANTITY']
      .transform('sum')
)
```

```python
df[['PRODUCT_NAME','BEST_SELLING']].sort_values(by='BEST_SELLING', ascending=False)
```

```python
df['COUNTRY_RANK'] = df.groupby('COUNTRY')['TOTAL_PRICE'].rank(method="dense", ascending=False)
```

```python
(df[['COUNTRY','COUNTRY_RANK']].sort_values(by='COUNTRY_RANK')).head(10)
```
